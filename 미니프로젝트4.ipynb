```python
# 업로드 된 모든 파일들을 디스크에 저장
import os

all_docs = []

# Save each uploaded file to disk and then load with PyPDFLoader
for filename, content in uploaded.items():
    # Write the content of the uploaded file to a local file
    with open(filename, 'wb') as f:
        f.write(content)
    print(f'Saved {filename} to disk.')

    # Load the PDF using PyPDFLoader
    loader = PyPDFLoader(filename)
    docs = loader.load()
    all_docs.extend(docs)

print(f'Total number of documents loaded: {len(all_docs)}')

"""
# Display the first few pages' content from the first loaded document as an example
if all_docs:
    print('\n--- Sample of extracted text (first 2 pages of the first document) ---')
    for i in range(min(2, len(all_docs))):
        print(f'Page {i+1}:\n{all_docs[i].page_content[:500]}...\n')
else:
    print('No documents were loaded.')
"""
```

```python

# 3. 머리글/바닥글/빈 페이지 제거
def clean_page(text):
    # 페이지 번호 제거 (단독으로 있는 숫자)
    text = re.sub(r'^\s*\d+\s*$', '', text, flags=re.MULTILINE)
    # PART 제거
    text = re.sub(r'PART\s*\d+', '', text)
    return text.strip()
```

```python
# 텍스트 정제 적용 및 빈 페이지 제거
cleaned_all_docs = []
for doc in all_docs:
    cleaned_content = clean_page(doc.page_content)
    if len(cleaned_content.strip()) > 100: # 100자 미만은 빈 페이지로 간주
        doc.page_content = cleaned_content
        cleaned_all_docs.append(doc)

print(f"정제 후 {len(cleaned_all_docs)} 페이지")

# 4. 장/절 구조 기준으로 쪼개기
splitter = RecursiveCharacterTextSplitter(
    separators=[
        r"\n제\d+장",   # 장 구분
        r"\n제\d+절",   # 절 구분
        r"\n\d+\.",     # 번호 목록 (1. 2. 3.)
        r"\n가\.",      # 가. 나. 다.
        r"\n\n",        # 문단
        r"\n",          # 줄바꿈
        r"."            # 문장
    ],
    chunk_size=1500,
    chunk_overlap=300,
    length_function=len,
    is_separator_regex=True
)

chunks = splitter.split_documents(cleaned_all_docs)
print(f"총 {len(chunks)}개 청크 생성")
```

```python
full_text_cleaned = " ".join([doc.page_content for doc in cleaned_all_docs])
print(f"클린 텍스트 총 길이: {len(full_text_cleaned)} 자")
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")
db = Chroma.from_documents(chunks, embeddings)
print("벡터 DB 저장 완료")
```

```python
#검색기 설정
retriever = db.as_retriever(search_kwargs={"k": 10})
```

### 공통 RAG 설정 및 유틸리티 함수

```python
rag_prompt_template = ChatPromptTemplate.from_template("""
아래 제공된 문서 블록들을 참고하여 질문에 답해 주세요. 각 문서 블록에는 내용과 함께 'Source' 및 'Page' 정보가 명시되어 있습니다.
답변을 생성할 때, 각 문서 블록에 명시된 'Source'와 'Page' 정보를 활용하여 답변의 출처를 정확히 표기해 주세요. 관련된 정보가 없으면 '문서에서 찾을 수 없습니다'라고 답해 주세요.

문서 내용:
{context}

질문: {question}
""")

def ask_rag(question):
    docs = retriever.invoke(question)
    formatted_docs = []
    for i, d in enumerate(docs):
        source_filename = d.metadata.get('source', 'Unknown Document')
        page_number = d.metadata.get('page', 'Unknown Page')
        formatted_docs.append(f"--- Document {i+1} ---\nSource: {source_filename}, Page: {page_number}\nContent:\n{d.page_content}")
    context = "\n\n".join(formatted_docs)
    chain = rag_prompt_template | llm | parser
    return chain.invoke({"context": context, "question": question})
```

### Summarize Extracted Text

Now that we have extracted text from all the PDF documents, we can use the `llm` (Gemini) model to summarize the content. We'll concatenate all the document content into a single string and then pass it to a summarization prompt.

```python
questions = [
    "이 문서들과 관련이 있는 테스트용 질의응답 데이터셋을 20개 생성해 주세요."
]

# 질문하기 (그림으로 된 내용 관련해서 답변은 어려움에 유의할 것)
for my_question in questions:
  print(ask_rag(my_question))
```

### 전체 문서 내용 요약

모든 PDF 문서에서 추출된 전체 텍스트(`full_text`)를 사용하여 LLM(Gemini) 모델로 전체 내용을 요약할 수 있습니다.

```python
summary_prompt = ChatPromptTemplate.from_template("""
다음 문서의 전체 내용을 명확하고 간결하게 요약해 주세요:

{text}
""")

summary_chain = summary_prompt | llm | parser

print("전체 문서 요약:")
print(summary_chain.invoke({"text": full_text_cleaned}))
```

### 특정 키워드 검색

전체 텍스트(`full_text`) 내에서 특정 키워드가 포함된 문장을 검색하는 간단한 방법입니다.

```python
keyword_to_search = "인형"

found_sentences = []
# Split full_text into sentences for better keyword search context
# This is a basic sentence split; for more robust splitting, consider NLTK or similar libraries.
sentences = re.split(r'(?<=[.!?]) +', full_text_cleaned)

for sentence in sentences:
    if keyword_to_search in sentence:
        found_sentences.append(sentence.strip())

print(f"\n키워드 '{keyword_to_search}' 검색 결과:")
if found_sentences:
    for i, sentence in enumerate(found_sentences[:5]): # Displaying up to 5 relevant sentences
        print(f"- {sentence}...")
    if len(found_sentences) > 5:
        print(f"... (총 {len(found_sentences)}개의 문장 중 처음 5개 표시)")
else:
    print(f"'{keyword_to_search}' 키워드를 찾을 수 없습니다.")
```
